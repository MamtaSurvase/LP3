```python
# Gradient Descent Algorithm

def function(x):
    return (x + 3) ** 2
```

```python
def gradient(x):
    return 2 * (x + 3)
```

```python

x = 2
learning_rate = 0.1
iterations = 50
```

```python
for i in range(iterations):
    x = x - learning_rate * gradient(x)

```

```python
print("Local minimum x =", round(x, 4))
print("Minimum value y =", round(function(x), 4))
```
